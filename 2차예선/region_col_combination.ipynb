```python
import pandas as pd
import numpy as np

train_1 = pd.read_csv('C:/Users/dpwl1/Downloads/data/data/train/train_1.csv')
train_2 = pd.read_csv('C:/Users/dpwl1/Downloads/data/data/train/train_2.csv')

meta_weather = pd.read_csv('C:/Users/dpwl1/Downloads/data/data/train/meta/TRAIN_기상_2018-2022.csv')

print("train_1 head:")
print(train_1.head())
print("\ntrain_2 head:")
print(train_2.head())
print("\nweather head:")
print(meta_weather.head())
```

```python
# 'YYYYMMSOON', '지역 이름', '주산지 품목명', '주산지 품종명' 제외
columns_to_exclude = ['YYYYMMSOON', '지역 이름', '주산지 품목명', '주산지 품종명']
weather_data = meta_weather.drop(columns=columns_to_exclude, errors='ignore')

# 주산지 품목명별로 먼저 그룹화한 후, 각 그룹 내에서 지역 이름별로 결측치가 없는 열을 찾음
result_dict = {}

for item_name, group_by_item in meta_weather.groupby('주산지 품목명'):
    region_columns_map = {}
    for region, group_by_region in group_by_item.groupby('지역 이름'):
        # 결측치가 없는 열 추출
        no_missing_columns = group_by_region[weather_data.columns].isna().sum() == 0
        region_columns_map[region] = list(weather_data.columns[no_missing_columns])
    
    # 주산지 품목명별로 지역과 결측치 없는 열을 저장
    result_dict[item_name] = region_columns_map

# 결과 출력
for item_name, region_dict in result_dict.items():
    print(f"주산지 품목명: {item_name}")
    for region, columns in region_dict.items():
        print(f"  지역: {region}, 결측치 없는 열: {columns}")
```

```python
# 'YYYYMMSOON', '지역 이름', '주산지 품목명', '주산지 품종명' 제외
columns_to_exclude = ['YYYYMMSOON', '지역 이름', '주산지 품목명', '주산지 품종명']
weather_data = meta_weather.drop(columns=columns_to_exclude, errors='ignore')

# 주산지 품목명별로 먼저 그룹화한 후, 각 그룹 내에서 지역 이름별로 결측치가 있는 열을 찾음
result_dict = {}

for item_name, group_by_item in meta_weather.groupby('주산지 품목명'):
    region_columns_map = {}
    for region, group_by_region in group_by_item.groupby('지역 이름'):
        # 결측치가 있는 열 추출
        missing_columns = group_by_region[weather_data.columns].isna().sum() > 0
        region_columns_map[region] = list(weather_data.columns[missing_columns])
    
    # 주산지 품목명별로 지역과 결측치가 있는 열을 저장
    result_dict[item_name] = region_columns_map

# 결과 출력
for item_name, region_dict in result_dict.items():
    print(f"주산지 품목명: {item_name}")
    for region, columns in region_dict.items():
        print(f"  지역: {region}, 결측치 있는 열: {columns}")
```
